```python
# Defining the Country House
# hypothesis is about country houses, but the dataset has no “country house” column. We have to create it by deciding which houses count. A country house has two features:

#1. It’s far from the city: you measure the distance from downtown Seattle.
#2. It has a large lot: you look at sqft_lot.
```

```python
#1. It’s far from the city: you measure the distance from downtown Seattle.
# Coordinates of downtown Seattle
SEA_LAT = 47.6062
SEA_LON = -122.3321
```

```python
import numpy as np
```

```python
# Convert degrees to radians (the formula needs radians)
lat1 = np.radians(df["lat"])
lon1 = np.radians(df["long"])
lat2 = np.radians(SEA_LAT)
lon2 = np.radians(SEA_LON)
```

```python
# Haversine formula: straight-line distance between two points on Earth
a = np.sin((lat2 - lat1) / 2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin((lon2 - lon1) / 2)**2
df["dist_miles"] = 3958.8 * 2 * np.arcsin(np.sqrt(a))   # 3958.8 = Earth's radius in miles
```

```python
#claude
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

# Check that 1/2 match how you built month_group
SEASONS = {
    1: {"label": "Autumn & Winter\n(Oct–Feb)", "color": "#3B6EA8"},
    2: {"label": "Spring & Summer\n(Mar–Sep)", "color": "#E08A3C"},
}
COLD, WARM = 1, 2

df_seasonal = seasonal_comparison.reset_index()          # columns: month_group, count, median
medians = df_seasonal.set_index("month_group")["median"]
cold, warm = medians[COLD], medians[WARM]
pct_diff = (cold - warm) / warm * 100
hypothesis_holds = cold < warm

# Title + recommendation are generated from the result
if hypothesis_holds:
    headline = f"Autumn & winter is {abs(pct_diff):.0f}% cheaper per sqft"
    advice = (f"Recommendation: buy between October and February - you pay about "
              f"${warm - cold:,.0f} less per sqft\nthan in spring/summer.")
else:
    headline = f"Autumn & winter is NOT cheaper ({pct_diff:+.0f}% per sqft)"
    advice = ("Recommendation: season does not lower the price - don't wait for Oct–Feb;\n"
              "buy when the right house is available.")

fig, ax = plt.subplots(figsize=(9, 6.5))
x = range(len(df_seasonal))
bars = ax.bar(x, df_seasonal["median"], width=0.55,
              color=[SEASONS[g]["color"] for g in df_seasonal["month_group"]])

# Value on top of each bar
for bar, val in zip(bars, df_seasonal["median"]):
    ax.text(bar.get_x() + bar.get_width() / 2, val + df_seasonal["median"].max() * 0.02,
            f"${val:,.0f}", ha="center", va="bottom", fontsize=15, fontweight="bold", color="#222")

# Season names + sample size as x labels (replaces the confusing count bars)
ax.set_xticks(list(x))
ax.set_xticklabels([f"{SEASONS[g]['label']}\n{int(n)} houses sold"
                    for g, n in zip(df_seasonal["month_group"], df_seasonal["count"])], fontsize=11)

# One clean y-axis, starting at 0
ax.set_ylabel("Median price per sqft (USD)", fontsize=11, color="#444")
ax.yaxis.set_major_formatter(mticker.StrMethodFormatter("${x:,.0f}"))
ax.set_ylim(0, df_seasonal["median"].max() * 1.18)
ax.grid(axis="y", color="#e5e5e5", linewidth=0.8)
ax.set_axisbelow(True)
for side in ("top", "right", "left"):
    ax.spines[side].set_visible(False)
ax.tick_params(axis="y", length=0, colors="#666")
ax.tick_params(axis="x", length=0)

# Headline, context, recommendation
fig.text(0.06, 0.965, headline, fontsize=16, fontweight="bold", ha="left", va="top")
fig.text(0.06, 0.915, "Non-renovated country houses, 400–2,000 sqft (William's criteria)",
         fontsize=11, color="#555", ha="left", va="top")
fig.text(0.06, 0.025, advice, fontsize=11.5, fontweight="bold",
         color="#1f5c2e" if hypothesis_holds else "#8a2a1f", ha="left", va="bottom")

fig.subplots_adjust(top=0.84, bottom=0.24, left=0.12, right=0.97)
plt.savefig("seasonal_comparison.png", dpi=200, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt
from scipy.stats import mannwhitneyu

# ---------- Data ----------
# (assumes both dataframes are already filtered to compact 1–2 bed houses)
center = final_city_center_houses["price_per_sqft"].dropna()
outer  = final_outer_houses["price_per_sqft"].dropna()

n_c, n_o = len(center), len(outer)
total = n_c + n_o
med_c, med_o = center.median(), outer.median()
premium = (med_c / med_o - 1) * 100

_, p = mannwhitneyu(center, outer, alternative="greater")
p_txt = "p < 0.001" if p < 0.001 else f"p = {p:.3f}"

C_CENTER, C_OUTER = "#1f5aa6", "#e08a2c"
labels = ["Within 10 mi\nof city centre", "Outer areas\n(> 10 mi)"]
colors = [C_CENTER, C_OUTER]

# ---------- Figure ----------
plt.rcParams.update({"font.size": 12, "axes.spines.top": False,
                     "axes.spines.right": False, "axes.spines.left": False})
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 6), gridspec_kw={"wspace": 0.3})

def clean(ax):
    ax.set_yticks([])          # values are printed on the bars instead
    ax.tick_params(axis="x", length=0)

# ---------- Panel A: how many houses ----------
counts = [n_c, n_o]
bars = ax1.bar(labels, counts, color=colors, width=0.55)
for bar, n in zip(bars, counts):
    ax1.text(bar.get_x() + bar.get_width() / 2, bar.get_height(),
             f"{n:,}\n({n / total:.0%})", ha="center", va="bottom",
             fontsize=14, fontweight="bold")
ax1.set_ylim(0, max(counts) * 1.3)
ax1.set_title("Number of compact 1–2 bed houses", fontweight="bold", pad=15)
clean(ax1)

# ---------- Panel B: price per sq ft ----------
medians = [med_c, med_o]
bars = ax2.bar(labels, medians, color=colors, width=0.55)
for bar, m in zip(bars, medians):
    ax2.text(bar.get_x() + bar.get_width() / 2, bar.get_height(),
             f"${m:,.0f}", ha="center", va="bottom",
             fontsize=14, fontweight="bold")
ax2.set_ylim(0, max(medians) * 1.45)
ax2.set_title("Median price per square foot", fontweight="bold", pad=15)
ax2.text(0.5, 0.92, f"+{premium:.0f}% in the city centre",
         transform=ax2.transAxes, ha="center", fontsize=15,
         fontweight="bold", color=C_CENTER)
ax2.text(0.5, 0.85, f"(statistically significant, {p_txt})",
         transform=ax2.transAxes, ha="center", fontsize=10, color="#555")
clean(ax2)

# ---------- Headline ----------
fig.suptitle("Compact 1–2 bed houses: more of them near the city centre, "
             "and they cost more per sq ft",
             fontsize=15, fontweight="bold", y=1.03)
plt.savefig("compact_houses_hypothesis.png", dpi=200, bbox_inches="tight")
plt.show()
```

```python
#Define final country houses those with 1 to 2 bedrooms an living space between 400 and 2000 sqft
final_country_houses_h3 = country_houses[(country_houses["bedrooms"].isin([1, 2])) & (country_houses["sqft_living"] >= 400) & (country_houses["sqft_living"] <= 2000)] 
```

```python
#define the unrenovated country houses as the final country houses with yr_renovated = 0
unrenovated_country_houses = final_country_houses_h3[final_country_houses_h3["yr_renovated"] == 0]
print("Number of unrenovated country houses:", unrenovated_country_houses.shape[0])  
```

```python
#define the renovated country houses as the final country houses with yr_renovated > 0
renovated_country_houses = final_country_houses_h3[final_country_houses_h3["yr_renovated"] > 0] 
print ("Number of renovated country houses:", renovated_country_houses.shape[0])
```

```python
#Price per square foot for unrenovated country houses and renovated country houses
unrenovated_country_houses["price_per_sqft"] = unrenovated_country_houses["price"] / unrenovated_country_houses["sqft_living"]
renovated_country_houses["price_per_sqft"] = renovated_country_houses["price"] / renovated_country_houses["sqft_living"]  
print("Price per square foot for unrenovated country houses:")
print(unrenovated_country_houses[["price_per_sqft"]].head())
print("\nPrice per square foot for renovated country houses:")
print(renovated_country_houses[["price_per_sqft"]].head())  
```

```python
#comparing the price per square foot for unrenovated country houses and renovated country houses
mean_price_per_sqft_unrenovated = unrenovated_country_houses["price_per_sqft"].mean()
mean_price_per_sqft_renovated = renovated_country_houses["price_per_sqft"].mean()
median_price_per_sqft_unrenovated = unrenovated_country_houses["price_per_sqft"].median()
median_price_per_sqft_renovated = renovated_country_houses["price_per_sqft"].median()
print(f"Mean price per square foot for unrenovated country houses: {mean_price_per_sqft_unrenovated:.2f}")
print(f"Mean price per square foot for renovated country houses: {mean_price_per_sqft_renovated:.2f}")
print(f"Median price per square foot for unrenovated country houses: {median_price_per_sqft_unrenovated:.2f}")
print(f"Median price per square foot for renovated country houses: {median_price_per_sqft_renovated:.2f}")     
```

```python
# Plot the price per square foot for unrenovated country houses and renovated country houses
plt.figure(figsize=(10, 6))
plt.hist(unrenovated_country_houses["price_per_sqft"], alpha=0.5, label="Unrenovated Country Houses")
plt.hist(renovated_country_houses["price_per_sqft"], alpha=0.5, label="Renovated Country Houses")
plt.xlabel("Price per Square Foot")
plt.ylabel("Frequency") 
```

```python
#calculation the difference between the price per square foot for unrenovated country houses and renovated country houses
price_diff = mean_price_per_sqft_renovated - mean_price_per_sqft_unrenovated
percent_diff = (price_diff / mean_price_per_sqft_unrenovated) * 100
print(f"Mean price per square foot for unrenovated country houses: ${mean_price_per_sqft_unrenovated:.2f}")
print(f"Mean price per square foot for renovated country houses: ${mean_price_per_sqft_renovated:.2f}")
print(f"Difference in mean price per square foot: ${price_diff:.2f} ({percent_diff:.2f}%)") 
```

```python
import matplotlib.pyplot as plt

unren = mean_price_per_sqft_unrenovated
ren = mean_price_per_sqft_renovated
cheaper = (ren - unren) / ren * 100   # % cheaper than renovated

fig, ax = plt.subplots(figsize=(7, 5.5))
bars = ax.bar(["Country houses\nnot renovated", "Country houses\nrenovated"],
              [unren, ren], color=["#d9822b", "#2b6cb0"], width=0.6)
ax.bar_label(bars, labels=[f"${unren:,.2f}", f"${ren:,.2f}"],
             fontsize=13, fontweight="bold", padding=4)
ax.annotate(f"{cheaper:.1f}% cheaper\n(${ren - unren:,.2f} less per sq ft)",
            xy=(0.2, unren * 1.08), xytext=(0.5, ren * 1.12),
            ha="center", fontsize=12, fontweight="bold", color="#d9822b",
            arrowprops=dict(arrowstyle="->", color="#d9822b", lw=2))

ax.set_ylim(0, ren * 1.3)
ax.set_yticks([])
ax.spines[["top", "right", "left"]].set_visible(False)
ax.tick_params(axis="x", labelsize=11)
ax.set_title("Average price per sq ft: country houses not renovated vs renovated",
             fontsize=14, fontweight="bold", pad=15)

plt.tight_layout()
plt.show()
```
